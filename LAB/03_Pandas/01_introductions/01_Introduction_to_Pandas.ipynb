## 1. Two main objects: `Series` and `DataFrame`

- **Series**: a one-dimensional array with an index, like one labeled column.
- **DataFrame**: a two-dimensional table, with rows and named columns.

NumPy arrays are ideal for numerical calculations; DataFrames are convenient for datasets combining numbers, text, dates and missing values.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Compare a NumPy array with a labeled Pandas Series
values = np.array([10, 20, 30])
series = pd.Series(values, name="sales")

print("NumPy array:", values)
print("Pandas Series:")
display(series)
```

## 2. Create a DataFrame

We'll use a **fictional campus café**. Each row is an order, and each column describes one property of that order. All data is included below, so the notebook works immediately in Colab.

```python
orders = pd.DataFrame({
    "date": ["2026-11-12", "2026-11-12", "2026-11-12", "2026-11-13",
             "2026-11-13", "2026-11-13", "2026-11-14", "2026-11-14",
             "2026-11-14", "2026-11-15", "2026-11-15", "2026-11-15"],
    "product": ["Espresso", "Sandwich", "Tea", "Cookie", "Espresso", "Sandwich",
                "Tea", "Cookie", "Sandwich", "Espresso", "Tea", "Cookie"],
    "category": ["Drink", "Food", "Drink", "Food", "Drink", "Food",
                 "Drink", "Food", "Food", "Drink", "Drink", "Food"],
    "quantity": [2, 1, 3, 2, 4, 2, 1, 3, 2, 3, 2, 1],
    "unit_price": [2.0, 6.5, 2.5, 1.8, 2.0, 6.5, 2.5, 1.8, 6.5, 2.0, 2.5, 1.8],
    "rating": [5, 4, 4, None, 5, 3, 4, None, 5, 4, 5, None]
})

display(orders)
```

## 3. Inspect data

When opening an unfamiliar dataset, begin with:

| Expression | Meaning |
|---|---|
| `df.head()` | First five rows |
| `df.shape` | Number of rows and columns |
| `df.columns` | Column names |
| `df.info()` | Types and non-missing values |
| `df.describe()` | Basic numeric statistics |

```python
display(orders.head())
print("Shape:", orders.shape)
print("Columns:", orders.columns.tolist())
orders.info()
display(orders.describe())
```

## 4. Select columns and filter rows

Use `df["column"]` for a Series, `df[["col1", "col2"]]` for a smaller DataFrame, and a **Boolean condition** to select certain rows.

```python
# One column
print("Products:")
display(orders["product"].head())

# Multiple columns
display(orders[["product", "quantity"]].head())

# Only orders with at least three units
display(orders[orders["quantity"] >= 3])

# Combine conditions with & (and) or | (or); use parentheses
display(orders[(orders["category"] == "Drink") & (orders["quantity"] >= 3)])
```

## 5. Create new columns and sort

Pandas applies calculations to **entire columns**, using vectorized operations like NumPy.

```python
orders["revenue"] = orders["quantity"] * orders["unit_price"]

# Highest-value orders first
display(orders.sort_values("revenue", ascending=False).head())

print("Total revenue:", orders["revenue"].sum())
print("Average quantity:", orders["quantity"].mean())
```

## 6. Group and summarize with `groupby()`

`groupby()` lets you answer questions such as **“How much revenue did each product generate?”** without writing a loop.

```python
revenue_by_product = orders.groupby("product")["revenue"].sum()
display(revenue_by_product.sort_values(ascending=False))

# You can also count orders per category
display(orders.groupby("category").size())
```

## 7. Missing data

A CSV may contain empty cells. Pandas represents many missing values as `NaN` (or `NA`, depending on the dtype).

- `isna()` detects missing entries.
- `fillna()` substitutes a value.
- `dropna()` removes entries with missing values.

**Be careful:** replacing a missing rating with 0 would mean "bad rating", not "unknown".

```python
print("Missing values per column:")
print(orders.isna().sum())

# Calculate an average from known ratings only
print("Average known rating:", orders["rating"].mean())

# Remove rows without ratings, without changing the original table
rated_orders = orders.dropna(subset=["rating"])
print("Rows with a rating:", len(rated_orders))
```

## 8. Reading and writing a CSV

Most real projects start from a CSV file. To keep this demonstration self-contained, we'll **save our own CSV first**, then reopen it with `read_csv()`. In Colab, the file is created in the runtime's current directory.

```python
orders.to_csv("campus_cafe_orders.csv", index=False)

loaded = pd.read_csv("campus_cafe_orders.csv")
display(loaded.head())

# Convert a text date column into actual dates when necessary:
loaded["date"] = pd.to_datetime(loaded["date"])
print("Date type:", loaded["date"].dtype)
```

## 9. Plot the results

Pandas provides a quick plotting interface using **Matplotlib** behind the scenes. You can also use `plt` directly for more control.

```python
revenue_by_product.sort_values(ascending=False).plot(kind="bar", figsize=(7, 4))
plt.title("Campus café — Revenue by product")
plt.xlabel("Product")
plt.ylabel("Revenue (€)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
