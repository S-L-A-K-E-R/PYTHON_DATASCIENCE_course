### Data prepared for you
This cell imports Matplotlib / NumPy and exposes the arrays you will use. **Run it once.**

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

# Data is supplied: no web downloads or data-cleaning assignments here.
DATA = Path('data')
if not DATA.exists():
    raise FileNotFoundError('Please open the notebook from the LAB folder containing data/ and figures/.')

air = np.genfromtxt(DATA / 'air_quality.csv', delimiter=',', names=True, encoding='utf-8')
model = np.genfromtxt(DATA / 'model_predictions.csv', delimiter=',', names=True, encoding='utf-8')
traffic = np.genfromtxt(DATA / 'traffic.csv', delimiter=',', names=True, encoding='utf-8')

hours = air['hour']
north = air['station_north_pm25']
south = air['station_south_pm25']
observed = model['observed']
predicted = model['predicted']
absolute_error = model['absolute_error']

traffic_hour = traffic['hour']
vehicles = traffic['vehicles']
speed = traffic['speed_kmh']
rain = traffic['rain_mm']
weekend = traffic['is_weekend'].astype(bool)
hour_labels = np.arange(24)
hourly_vehicles = np.array([vehicles[traffic_hour == h].mean() for h in hour_labels])
daytype_labels = ['Weekdays', 'Weekends']
daytype_vehicles = [vehicles[~weekend].mean(), vehicles[weekend].mean()]

print('Ready: air-quality time series, model predictions and traffic arrays.')
print(f'Air quality: {len(hours)} hours | ML observations: {len(observed)} | Traffic readings: {len(vehicles)}')
```
