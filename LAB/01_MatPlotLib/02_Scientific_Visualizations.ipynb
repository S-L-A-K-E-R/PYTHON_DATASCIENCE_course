### Prepared data
Run once. Data loading and the 2D mesh are provided.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

DATA = Path('data')
if not DATA.exists():
    raise FileNotFoundError('Please open the notebook from the LAB folder containing data/ and figures/.')

calibration = np.genfromtxt(DATA / 'calibration.csv', delimiter=',', names=True, encoding='utf-8')
reference = calibration['reference_c']
measured = calibration['measured_c']
uncertainty = calibration['uncertainty_sd_c']

field = np.load(DATA / 'server_room_grid.npz')
x = field['x']      # length of room: 0...12 m
y = field['y']      # width of room: 0...8 m
temperature = field['temperature']
X, Y = np.meshgrid(x, y)

print('Ready: calibration points and a 2D thermal grid.')
print('Grid shapes:', X.shape, Y.shape, temperature.shape)
```
