```python
from pathlib import Path
from urllib.request import urlretrieve

import numpy as np
import matplotlib.pyplot as plt

# Teacher-provided file access. Students only need to use the loaded arrays.
def data_file(filename):
    possible_paths = (
        Path("data") / filename,
        Path("LAB/01_MatPlotLib/data") / filename,
    )
    for candidate in possible_paths:
        if candidate.is_file():
            return candidate

    destination = Path("data") / filename
    destination.parent.mkdir(parents=True, exist_ok=True)
    base_url = ("https://raw.githubusercontent.com/S-L-A-K-E-R/"
                "PYTHON_DATASCIENCE_course/master/LAB/01_MatPlotLib/data/")
    try:
        urlretrieve(base_url + filename, destination)
    except Exception as exc:
        raise FileNotFoundError(
            "Dataset not found locally and automatic download failed. "
            "Connect to the internet or keep the LAB/01_MatPlotLib/data folder."
        ) from exc
    return destination

print("Matplotlib and NumPy are ready.")
```
