### 1 — A Python list is not a NumPy array

Predict the results. What does `* 2` mean for each type?


````python
import numpy as np

python_list = [2, 4, 6]
numpy_array = np.array([2, 4, 6])

print("Python list:", python_list * 2)
print("NumPy array:", numpy_array * 2)

# Try changing the multiplier from 2 to 3.
````

**Expected result at first run:**

```text
Python list: [2, 4, 6, 2, 4, 6]
NumPy array: [ 4  8 12]
```

For NumPy arrays, ordinary arithmetic can act **element by element**. For Python lists, multiplying by an integer repeats the list.


### 2 — A matrix: rows and columns

In NumPy, a 2D array can represent a matrix. Each inner list becomes **one row**. Indexing starts at 0, just as in Python lists.

Predict the shape and the value in row 1, column 2.


````python
room = np.array([
    [20, 22, 24],
    [22, 24, 26],
])

print("Matrix:\n", room)
print("Shape:", room.shape)
print("Value at [1, 2]:", room[1, 2])
````

**Expected:**

```text
Shape: (2, 3)
Value at [1, 2]: 26
```

A shape of **(2, 3)** means **2 rows and 3 columns**. `room[1, 2]` selects the second row, third column.


### 3 — Make all values change

The room thermometer reads 2°C too low. We can add an offset to every cell **without writing a loop**. Try it, then modify the offset to `-1`.


````python
offset = 2
corrected = room + offset

print(corrected)

# Experiment: set offset to -1, then run this cell again.
````

**Expected for `offset = 2`:**

```text
[[22 24 26]
 [24 26 28]]
```

**Expected for `offset = -1`:**

```text
[[19 21 23]
 [21 23 25]]
```

This is an example of **broadcasting with a scalar**: the same number is applied to each element. We will explore more broadcasting rules another day.


### 4 — Summarize rows or columns

`np.mean` calculates averages. `axis=0` returns one average **per column**; `axis=1` returns one **per row**.


````python
print("Whole matrix:", np.mean(room))
print("By column:", np.mean(room, axis=0))
print("By row:", np.mean(room, axis=1))
````

**Expected:**

```text
Whole matrix: 23.0
By column: [21. 23. 25.]
By row: [22. 24.]
```
