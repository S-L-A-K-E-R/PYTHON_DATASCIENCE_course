### 0 — How an RGB image becomes an array

An RGB pixel has three integer values: **[red, green, blue]**, usually between **0 and 255**.

A color image with **H rows** and **W columns** has shape **`(H, W, 3)`**.

> **Coordinates:** `image[row, column]` or `image[y, x]`. The **row comes first**, unlike the usual `x, y` order in mathematics.

First, run this tiny example. It is a **2-row × 3-column** image with six colored pixels.

````python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import matplotlib.cbook as cbook

toy = np.array([
    [[255, 0, 0], [0, 255, 0], [0, 0, 255]],
    [[255, 255, 0], [0, 255, 255], [255, 0, 255]],
], dtype=np.uint8)

print("Shape:", toy.shape)
print("Pixel at row 0, column 0:", toy[0, 0])
plt.imshow(toy, interpolation="nearest")
plt.title("Six RGB pixels")
plt.xticks(range(3))
plt.yticks(range(2))
plt.show()
````
