## Setup — RGB pixels

Each RGB pixel has three values, ordered **red, green, blue**. Images use `image[row, column]` (also written `image[y, x]`).

Run the small example and the Lenna photograph setup before working on the exercises.

**RGB reference:**

<img src="https://raw.githubusercontent.com/S-L-A-K-E-R/PYTHON_DATASCIENCE_course/master/LAB/02_NumPy/assets/lab4/rgb_legend.jpg" alt="Red, green, blue channels and their combined RGB representation" width="440">

````python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from io import BytesIO
from urllib.request import urlopen

toy = np.array([
    [[255, 0, 0], [0, 255, 0], [0, 0, 255]],
    [[255, 255, 0], [0, 255, 255], [255, 0, 255]],
], dtype=np.uint8)

print("Toy array shape:", toy.shape)
plt.imshow(toy, interpolation="nearest")
plt.title("Six RGB pixels")
plt.xticks(range(3))
plt.yticks(range(2))
plt.show()

# Classic 512 x 512 RGB test photograph (downloaded automatically).
image_url = "https://raw.githubusercontent.com/mikolalysenko/lena/master/lena.png"
with urlopen(image_url, timeout=30) as response:
    image = np.asarray(Image.open(BytesIO(response.read())).convert("RGB"))

print("Lenna loaded:", image.shape, image.dtype)
plt.figure(figsize=(5, 5))
plt.imshow(image)
plt.title("Lenna — Original photograph")
plt.axis("off")
plt.show()
````
